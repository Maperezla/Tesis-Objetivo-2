```python
import numpy as np
import rasterio
import os

QUEMADO_BINARIO = r"D:\Maestria_Geomatica\Semestre_IV\Tesis_Python\Datos\TIFF\outputs_inferencia_ModelA_26_09_14\rf_quemado_binario_26_09_14.tif"

OUT_DIR = r"D:\Maestria_Geomatica\Semestre_IV\Tesis_Python\Datos\TIFF\outputs_inferencia_ModelA_26_09_14"
os.makedirs(OUT_DIR, exist_ok=True)

OUT_FILTRADO = os.path.join(OUT_DIR, "QUEMADO_BINARIO_FILTRADO_26_09_17.tif")

# 1. Abrir el raster original y copiar su metadata geográfica
with rasterio.open(QUEMADO_BINARIO) as src:
    raster_data = src.read(1)
    meta = src.meta.copy()
```

```python
    # 2. Reemplazar los valores que no sean 1 por un valor NoData (por ejemplo, 0 o 255)
    nodata_val = 0
    nuevo_raster_data = np.where(raster_data == 1, 1, nodata_val)

    # 3. Actualizar la metadata con el valor NoData asignado
    meta.update(
        nodata=nodata_val
    )

    # 4. Crear y escribir el nuevo archivo raster
    with rasterio.open(OUT_FILTRADO, "w", **meta) as dst:
        dst.write(nuevo_raster_data, 1)
```
